```python
import os, glob

train_splits = glob.glob("/kaggle/input/**/ImageSets/Segmentation/train.txt", recursive=True)
if not train_splits:
    raise FileNotFoundError("VOC2012 Segmentation files not found in /kaggle/input.")

voc2012_dir = os.path.dirname(os.path.dirname(os.path.dirname(train_splits[0])))
os.makedirs("data/VOCdevkit", exist_ok=True)
target_link = "data/VOCdevkit/VOC2012"

if os.path.islink(target_link) or os.path.exists(target_link):
    os.unlink(target_link)

os.symlink(voc2012_dir, target_link)
print(f"Successfully linked: {voc2012_dir} -> {target_link}")
```
